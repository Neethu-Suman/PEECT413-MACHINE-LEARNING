#Simple Example for K means

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

X = np.array([ [1, 2],  [2, 1],     [4, 5],     [5, 4],     [6, 5],     [7, 6] ])

# Define the initial centroids as specified
initial_centroids = np.array([[1, 2], [5, 4]])
k = len(initial_centroids)

# Create a copy of the initial centroids for plotting
initial_centroids_plot = np.copy(initial_centroids)
```

##Plotting the Initial K-Means State

```python
# Plot initial data points and centroids
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c='blue', marker='o', label='Data Points')
plt.scatter(initial_centroids_plot[:, 0], initial_centroids_plot[:, 1], c='red', marker='X', s=200, label='Initial Centroids')
plt.title('Initial K-Means State')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```

##Applying K-Means with specified initial centroids

```python
# Initialize KMeans with the specified initial centroids
kmeans = KMeans(n_clusters=k, init=initial_centroids, n_init=1, random_state=0)

# Fit the model and predict clusters
kmeans.fit(X)
y_pred = kmeans.predict(X)

# Get the final centroids
final_centroids = kmeans.cluster_centers_
```

##Plotting the Final K-Means State

```python
# Plot the clustered data points and final centroids
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='viridis', marker='o', label='Clustered Data Points')
plt.scatter(final_centroids[:, 0], final_centroids[:, 1], c='red', marker='X', s=200, label='Final Centroids')
plt.title('Final K-Means State')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```
